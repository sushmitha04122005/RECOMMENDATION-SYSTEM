```python
from imblearn.over_sampling import SMOTE
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop(columns=['Class','Time','Amount'],axis=1)
y=df['Class']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=24)
```

```python
smote=SMOTE(random_state=42)
x_trainsmp,y_trainsmp=smote.fit_resample(x_train,y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report
```

```python
decision_tree=DecisionTreeClassifier()
random_forest=RandomForestClassifier()
```

```python
decision_tree.fit(x_trainsmp,y_trainsmp)
```

```python
dt_pred=decision_tree.predict(x_test)
```

```python
dt_accuracy=accuracy_score(y_test,dt_pred)
print(f"Decision Tree Accuracy: {dt_accuracy}")
```

```python
print(classification_report(y_test,dt_pred))
```

```python
random_forest.fit(x_trainsmp,y_trainsmp)
```

```python
rf_pred=random_forest.predict(x_test)
```

```python
acc=accuracy_score(y_test,rf_pred)
print(f"accuracy_score :{acc}")
```

```python
classi=classification_report(y_test,rf_pred)
print(f"classification report/n:{classi}")
```
